# Earthing & Lightning Study — Musina 33/11 kV Substation
**Brownfield Protection Review**  
**Standards: IEEE 80, SANS 10313, IEC 62305**

```python
import numpy as np

# Soil and grid parameters (from DESIGN_BASIS.md)
rho_soil   = 120       # ohm·m
A_grid     = 40 * 30   # 1200 m²
L_total    = (9*30) + (7*40)  # 270 + 280 = 550 m (conductor length)
d_conductor= 0.0056    # m (25 mm²)
depth      = 0.5       # m

# Sverak formula (IEEE 80 Eq. 57)
Rg = rho_soil * (1/L_total + 1/np.sqrt(20*A_grid) * (1 + 1/(1 + 0.5*np.sqrt(20/A_grid))))
print(f"Estimated earth grid resistance: {Rg:.3f} ohm")

# Ground potential rise for 300 A earth fault
I_ef = 300.0
GPR = I_ef * Rg
print(f"GPR for 300 A earth fault: {GPR:.1f} V")

# Tolerable touch/step voltages (70 kg person, 0.5 s clearing)
rho_surface = 3000     # ohm·m (crushed stone)
ts = 0.5               # s (relay + breaker clearing time)
E_touch = (1000 + 1.5*rho_surface) * 0.116 / np.sqrt(ts)
E_step  = (1000 + 6.0*rho_surface) * 0.116 / np.sqrt(ts)
print(f"\nIEEE 80 tolerable touch voltage: {E_touch:.0f} V")
print(f"IEEE 80 tolerable step voltage:  {E_step:.0f} V")

# Approximate mesh and step voltages (grid factors)
Em = GPR * 0.65   # rough mesh voltage factor
Es = GPR * 0.40   # rough step voltage factor
print(f"\nEstimated mesh (touch) voltage: {Em:.1f} V")
print(f"Estimated step voltage:         {Es:.1f} V")
print(f"Touch voltage compliant: {Em < E_touch}")
print(f"Step voltage compliant:  {Es < E_step}")
if Em > E_touch:
    print("\n*** DEFICIENCY D‑006: Touch potential exceeds IEEE 80 tolerable limit ***")
    print("    Earthing grid upgrade required.")
```

```python
# Ground flash density (Limpopo)
Ng = 12.0               # flashes/km²/year
A_collection = 0.04     # km² (substation collection area)
N_strikes = Ng * A_collection
print(f"Expected lightning strikes per year: {N_strikes:.1f}")

# Arrester parameters (33 kV, Class 2)
V_res = 85e3    # residual voltage at 10 kA (V)
I_surge = 10e3  # A
t_surge = 50e-6 # s (8/20 µs waveshape)
E_surge = V_res * I_surge * t_surge   # joules
E_rating = 2500  # J (typical class 2)
print(f"\nEnergy per 10 kA surge: {E_surge:.0f} J")
print(f"Arrester energy rating (Class 2): {E_rating} J")
print(f"Single‑surge margin: {E_rating/E_surge:.1f} ×")

# With poor earthing, surge voltage is higher due to reflected waves
# If grid impedance is high, arrester must absorb more energy.
# Show the effect if grid impedance doubles the effective surge current.
I_surge_poor = 15e3   # higher surge due to reflections
E_surge_poor = V_res * I_surge_poor * t_surge
print(f"\nWith poor earthing (higher surge current 15 kA): {E_surge_poor:.0f} J")
print(f"Margin: {E_rating/E_surge_poor:.1f} × → thermal stress increases.")
print("This is the mechanism causing annual arrester failures (D‑007).")
```
